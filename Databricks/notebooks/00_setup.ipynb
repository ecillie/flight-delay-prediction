```python

CATALOG = "workspace"

SCHEMAS = {
    "raw": "default",
    "bronze": "bronze",
    "silver": "silver",
    "gold": "gold",
}


RAW_VOLUME = f"/Volumes/{CATALOG}/{SCHEMAS['raw']}/raw_data"


DATASETS = {
    "flights": f"{RAW_VOLUME}/flights",
    "weather": f"{RAW_VOLUME}/weather",
    "airports": f"{RAW_VOLUME}/airports",
}

print(f"Catalog: {CATALOG}")
print(f"Raw volume: {RAW_VOLUME}")
```

```python
for layer in ["bronze", "silver", "gold"]:
    schema = f"{CATALOG}.{SCHEMAS[layer]}"

    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {schema}")

    print(f"Schema ready: {schema}")
```

```python
display(
    spark.sql(f"SHOW SCHEMAS IN {CATALOG}")
)
```

```python
for name, path in DATASETS.items():
    try:
        files = dbutils.fs.ls(path)

        print(f"[OK] {name}")
        print(f"Path: {path}")
        print(f"Top-level entries: {len(files)}")

    except Exception as e:
        print(f"[ERROR] {name}: {e}")
```
